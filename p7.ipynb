```python
import time

def coin_change(coins, amount):
    # dp[i] = minimum number of coins needed to make amount i
    dp = [float('inf')] * (amount + 1)

    # 0 coins are needed to make amount 0
    dp[0] = 0

    # Calculate minimum coins for each amount
    for i in range(1, amount + 1):
        for coin in coins:
            if coin <= i:
                dp[i] = min(dp[i], dp[i - coin] + 1)

    # If amount cannot be formed
    if dp[amount] == float('inf'):
        return -1

    return dp[amount]


# User input
coins = list(map(int, input("Enter coin denominations separated by spaces: ").split()))
amount = int(input("Enter the amount: "))

print("\nCoins:", coins)
print("Amount:", amount)

# Start execution time
start_time = time.perf_counter()

# Find minimum number of coins
result = coin_change(coins, amount)

# End execution time
end_time = time.perf_counter()

# Calculate execution time
execution_time = end_time - start_time

# Display result
if result == -1:
    print("The amount cannot be formed using the given coins.")
else:
    print("Minimum number of coins:", result)

print(f"Execution time: {execution_time:.10f} seconds")

# Time complexity
print("\nTime Complexity: O(n × A)")
print("Space Complexity: O(A)")
print("where n = number of coin denominations")
print("and A = target amount")
```
